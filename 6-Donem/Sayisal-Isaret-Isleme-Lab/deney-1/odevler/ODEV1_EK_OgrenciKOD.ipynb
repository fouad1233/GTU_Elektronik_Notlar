<h4> Soru 1) 𝑎=3−4𝑗 ve 𝑏=1+2𝑗 olmaküzere 𝐶=𝑎+𝑏’dir.</h4><br>
𝐶 kompleks sayısının <br>
I. gerçek,<br>
II. sanal,<br>
III. genlik ve<br>
IV. faz<br>
bileşenlerini print() komutu ile bastırın.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = 3-4j
b = 1+2j
C = a + b
C_real = C.real
C_imag = C.imag
C_abs = np.abs(C)
C_angle = np.angle(C)
print("C sayısının gerçek kısmı = ", C_real)
print("C sayısının sanal kısmı = ", C_imag)
print("C sayısının mutlak değeri = ", C_abs)
print("C sayısının açısı = ", C_angle)
```

<h4>Soru 2)Aşağıdaki işaretlerin periyodunu bulun ve 2 tam periyot olacak şekilde çizdirin. Çizdiğiniz grafiğe bakarak periyodu teyit edin. </h4><br>

I.  𝑥[𝑛] = ⁡2sin⁡( (2𝜋/5)𝑛 − 𝜋/5)<br>
II. 𝑥[𝑛] = ⁡sin⁡((3𝜋/5)𝑛 − 2𝜋/5)<br> 

<h5> 2)I.)𝑥[𝑛] = ⁡2sin⁡( (2𝜋/5)𝑛 − 𝜋/5) işaretinin periyodunu hesaplayalım</h5><br>
işaretin faz gecikmesiz genliğe normalize edilmiş hali bu şekilde --> ⁡sin⁡((2𝜋/5)𝑛)<br>
⁡sin⁡((2𝜋/5)𝑛) --> w0 = (2𝜋/5)<br>
Periyodu bulmak için: 2𝜋K = w0*N -->  2𝜋K = (2𝜋/5)*N --> N=5K , K=1 için N0 = 5 işaretin temel periyodudur.<br>
Aşağıda işareti çizip gözlemleyelim:

```python
N0 = 5
n = np.arange(0, 2*N0)
#𝑥[𝑛] = ⁡2sin⁡( (2𝜋/5)𝑛 − 𝜋/5)
x = 2*np.sin((2*np.pi/5)*n - np.pi/5)

plt.stem(n, x)
plt.title(" x[n] = 2sin((2*pi/5)n - pi/5)")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

<h5>2) II. 𝑥[𝑛] = ⁡sin⁡((3𝜋/5)𝑛 − 2𝜋/5) işaretinin periyodunu hesaplayalım</h5>
işaretin faz gecikmesiz genliğe normalize edilmiş hali bu şekilde --> ⁡sin⁡((3𝜋/5)𝑛)<br>
sin((3𝜋/5)𝑛) --> w0 = (3𝜋/5)<br>
İşaretin periyodunu bulmak için:<br>
2𝜋K = w0N --> 2𝜋K = (3𝜋/5)N --> N = 10K/3 --> K=3 için N0 = 10 İşaretinin temel periyodudur

```python
N0 = 10
n = np.arange(0, 2*N0)
#𝑥[𝑛] = ⁡sin⁡((3𝜋/5)𝑛 − 2𝜋/5)
x = np.sin((3*np.pi/5)*n - 2*np.pi/5)
plt.stem(n, x)
plt.title(" x[n] = sin((3*pi/5)n - 2*pi/5)")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

<h4>Soru 3) Aşağıdaki işaretleri belirtilen aralıkta çizdiriniz, karşılaştırınız ve yorumlayınız. İndis vektörü
ile çiziminizde yatay eksen değerleri uyuşmalıdır. Grafiğin başlık kısmında çizdirilen fonksiyonun adı
(𝑥1[𝑛],⁡𝑥2[𝑛], vb.) belirtilmelidir.</h4><br>

I.𝑥1[𝑛]=𝑒^(𝜋/5) ,𝑛=0,1,...,15 <br>
II.𝑥3[𝑛]=𝑒^((j𝜋/5)n) ,𝑛=0,1,...,15 <br>
III.𝑥4[𝑛]=cos(𝜋/5)+𝑗sin(𝜋/5) ,𝑛=0,1,...,15 <br>

V.𝑥6[𝑛]=cos( (𝜋/5) 𝑛)+𝑗sin( (𝜋/5) 𝑛),𝑛=−10,−9,...,3,4 <br>


<h5>3)I.𝑥1[𝑛]=𝑒^(𝜋/5) ,𝑛=0,1,...,15  işaretini çizelim </h5>

```python
n = np.arange(0, 16)
n_ones = np.ones(16)
x = np.exp((np.pi/5)*n_ones)
plt.stem(n, x)
plt.title(" x[n] = e^((pi/5))")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

Yukarda işaret çizilmiştir. İşaret n den bağımsızdır bu yüzden işaret sabittir. Aynı zamanda işaret real yani gerçektir, sanal kısmı yoktur.

<h5>3) II. 𝑥3[𝑛]=𝑒^((j𝜋/5)n) ,𝑛=0,1,...,15 işaretini çizelim </h5>

```python
n = np.arange(0, 16)
x = np.exp((1j*np.pi/5)*n)
x_real = x.real
x_imag = x.imag
fig = plt.figure()
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
ax1.stem(n, x_real)
ax1.set_title("x[n] = e^((j*pi/5)n) işaretinin Gerçek kısım")
ax1.set_xlabel("n")
ax1.set_ylabel("Re{x[n]}")
ax2.stem(n, x_imag)
ax2.set_title("x[n] = e^((j*pi/5)n) işaretinin Sanal kısım")
ax2.set_xlabel("n")
ax2.set_ylabel("Im{x[n]}")
plt.subplots_adjust(hspace=1.)
plt.show()
```

Yukarda işaret çizilmiştir. işaretin gerçek ve sanal kısımları bulunmaktadır. İşaretin real ve sanal kısımları periyodiktir ve periyotları 10'dur.

<h5>3) III.𝑥4[𝑛]=cos(𝜋/5)+𝑗sin(𝜋/5) ,𝑛=0,1,...,15 işaretini çizelim</h5>

```python
n = np.arange(0, 16)
n_ones = np.ones(16)
#𝑥4[𝑛]=cos(𝜋/5)+𝑗sin(𝜋/5)
x = (np.cos(np.pi/5) + 1j*np.sin(np.pi/5))*n_ones
x_real = x.real
x_imag = x.imag
fig = plt.figure()
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
ax1.stem(n, x_real)
ax1.set_title("x4[n] = cos(pi/5) + jsin(pi/5)  işaretinin Gerçek kısım")
ax1.set_xlabel("n")
ax1.set_ylabel("Re{x4[n]}")
ax2.stem(n, x_imag)
ax2.set_title("x4[n] = cos(pi/5) + jsin(pi/5) işaretinin Sanal kısım")
ax2.set_xlabel("n")
ax2.set_ylabel("Im{x4[n]}")
plt.subplots_adjust(hspace=1.)
plt.show()
```

Yukarıda işaret çizilmiştir. İşaret n'e bağımlı değildir bu yüzden sabittir. İşaretin hem gerçek hemde sanal kısmı vardır.

<h5> 3) IV.𝑥5[𝑛] = cos ( (𝜋/5) 𝑛) + 𝑗sin( (𝜋/5) 𝑛) , 𝑛 = 0,1, ... ,15  işaretini çizdirelim</h5>

```python
n = np.arange(0, 16)
#𝑥5[𝑛] = cos ( (𝜋/5) 𝑛) + 𝑗sin( (𝜋/5) 𝑛)
x = np.cos((np.pi/5)*n) + 1j*np.sin((np.pi/5)*n)
x_real = x.real
x_imag = x.imag
fig = plt.figure()
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
ax1.stem(n, x_real)
ax1.set_title("x5[n] = cos((pi/5)n) + jsin((pi/5)n) işaretinin Gerçek kısımı")
ax1.set_xlabel("n")
ax1.set_ylabel("Re{x5[n]}")
ax2.stem(n, x_imag)
ax2.set_title("x5[n] = cos((pi/5)n) + jsin((pi/5)n) işaretinin Sanal kısımı")
ax2.set_xlabel("n")
ax2.set_ylabel("Im{x5[n]}")
plt.subplots_adjust(hspace=1.)
plt.show()
```

Bu işaret II. deki işaret ile aynıdır. Euler açılımından elde edilmiştir. Gerçek ve sanal kısımları vardır. Hem gerçek hem sanal kısmının periyodu 10'dur. Bu yüzden işaretin periyodu de 10'dur.

<h5>V.𝑥6[𝑛]=cos( (𝜋/5) 𝑛)+𝑗sin( (𝜋/5) 𝑛),𝑛=−10,−9,...,3,4 işaretini çizelim </h5>

```python
n = np.arange(-10, 5)
#𝑥6[𝑛]=cos( (𝜋/5) 𝑛)+𝑗sin( (𝜋/5) 𝑛)
x = np.cos((np.pi/5)*n) + 1j*np.sin((np.pi/5)*n)
x_real = x.real
x_imag = x.imag
fig = plt.figure()
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
ax1.stem(n, x_real)
ax1.set_title("x6[n] = cos((pi/5)n) + jsin((pi/5)n) işaretinin Gerçek kısımı")
ax1.set_xlabel("n")
ax1.set_ylabel("Re{x6[n]}")
ax2.stem(n, x_imag)
ax2.set_title("x6[n] = cos((pi/5)n) + jsin((pi/5)n) işaretinin Sanal kısımı")
ax2.set_xlabel("n")
ax2.set_ylabel("Im{x6[n]}")
plt.subplots_adjust(hspace=1.)
plt.show()
```

Bu işaretin hem gerçek hem sanal kısmı vardır. İkisi de 10 ile periyodiktir bu yüzden işaretin periyodu de 10'dur.

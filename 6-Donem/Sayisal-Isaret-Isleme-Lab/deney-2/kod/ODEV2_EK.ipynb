## Soru 1

$
\begin{aligned}
& Y\left(e^{j \omega}\right)-1,2 e^{-j \omega} Y\left(e^{j \omega}\right)=X\left(e^{j \omega}\right)-e^{-j 4 \omega} X\left(e^{j \omega}\right) \\
& \left(1-1,2 e^{-j \omega}\right) Y\left(e^{j \omega}\right)=\left(1-e^{-j \omega 4}\right) X\left(e^{j \omega}\right) \\
& H\left(e^{j \omega}\right)=\frac{Y\left(e^{j \omega}\right)}{X\left(e^{j \omega}\right)}=\frac{\left(1-e^{-j \omega 4}\right)}{\left(1-1,2 e^{-j \omega}\right)}=\frac{1}{1-1,2 e^{-j \omega}}-\frac{e^{-j \omega 4}}{1-1,2 e^{-j \omega}} \\
& h[n]=(1,2)^{n} u[n]-(1,2)^{n-4} u[n-4] \\
\end{aligned}
$

$\sum_{-\infty}^{\infty}|h[n]|<\infty$ ise sistem kararlidir
$$
\begin{aligned}
& \sum_{-\infty}^{\infty}|h[n]|=\sum_0^{\infty}|h[n]|= \\
& (1,2)^0+(1,2)^1+(1,2)^2+(1,2)^3+\sum_{n=4}^{\infty}(1,2)^n-(1,2)^n \cdot(1,2)^{-4} \\
& \text { ve } \sum_{n=4}^{\infty}(1,2)^n-(1,2)^n \cdot(1,2)^{-4}=\sum_{n=4}^{\infty}(1,2)^n\left[1-(1,2)^{-4}\right]
\end{aligned}
$$
ve $\sum_{n=4}^{\infty}(1,2)^n$ sonusuza gittiginden sistem kararsızdır

## Soru 2
Aşağıdaki sisteme (a) ve (b) şıklarında verilen giriş işaretleri uygulandığında bu iki sistemin çıkışında elde edilecek işaretleri bulunuz. Hem giriş işaretinin hem de çıkış işaretinin grafiklerini çizdiriniz. Not: Toplam 2 çıkış işareti elde edeceksiniz.

SİSTEM-1: h1[𝑛] = (1/3) (𝛿[𝑛 + 1] + 𝛿[𝑛] + 𝛿[𝑛 − 1]) <br>
a) 𝑥1[𝑛]=cos((𝜋/5)𝑛)(𝑢[𝑛−10]−𝑢[𝑛−20])  <br>
b) 𝑥2[𝑛] = (−0.3)𝑛(𝑢[𝑛 − 3] − 𝑢[𝑛 − 5])<br>

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

### a)
h1[𝑛] = (1/3) (𝛿[𝑛 + 1] + 𝛿[𝑛] + 𝛿[𝑛 − 1]) <br>
𝑥1[𝑛]=cos((𝜋/5)𝑛)(𝑢[𝑛−10]−𝑢[𝑛−20])  <br>

```python
def u(n):
    return np.where(n >= 0, 1, 0)
def delta(n):
    return np.where(n == 0, 1, 0)

n = np.arange(-20,21)
h1n = (1/3)*(delta(n+1) + delta(n)+ delta(n-1))
#print(h1n)
#𝑥1[𝑛]=cos((𝜋/5)𝑛)(𝑢[𝑛−10]−𝑢[𝑛−20])
x1n = np.cos((np.pi/5)*n)*(u(n-10)-u(n-20))
#print(x1n)

y1n = signal.convolve(x1n, h1n,"same")#[abs(nmin -nXmin):abs(nmin - nXmax)]
#print(y1n)
#plot x4n and y4n with subplot
plt.subplot(2, 1, 1)
plt.stem(n, x1n)
plt.title('x1[n]')
plt.xlabel('n')
plt.ylabel('x1[n]')
plt.subplots_adjust(hspace=0.8)
plt.grid()
plt.subplot(2, 1, 2)
#print(y1n)
#print(len(y1n))
plt.stem(n,y1n)
plt.title('y1[n]')
plt.xlabel('n')
plt.ylabel('y1[n]')
plt.grid()
plt.show()
```

### b)
h1[𝑛] = (1/3) (𝛿[𝑛 + 1] + 𝛿[𝑛] + 𝛿[𝑛 − 1]) <br>
𝑥2[𝑛] = (−0.3)𝑛(𝑢[𝑛 − 3] − 𝑢[𝑛 − 5])<br>

```python
def u(n):
    return np.where(n >= 0, 1, 0)
def delta(n):
    return np.where(n == 0, 1, 0)

n = np.arange(-10,11)
h1n = (1/3)*(delta(n+1) + delta(n)+ delta(n-1))
#print(h1n)
#𝑥2[𝑛] = (−0.3)𝑛(𝑢[𝑛 − 3] − 𝑢[𝑛 − 5])
x2n = ((-0.3)**n)*(u(n-3)-u(n-5))
#print(x2n)
y2n = signal.convolve(x2n, h1n,mode = "same")
#print(y2n)
#plot x4n and y4n with subplot
plt.subplot(2, 1, 1)
plt.stem(n, x2n)
plt.title('x2[n]')
plt.xlabel('n')
plt.ylabel('x2[n]')
plt.subplots_adjust(hspace=0.8)
plt.grid()
plt.subplot(2, 1, 2)
plt.stem(n,y2n)
plt.title('y2[n]')
plt.xlabel('n')
plt.ylabel('y2[n]')
plt.grid()
plt.show()

```

### c)
<br>
𝑦[𝑛] − 0.8𝑦[𝑛 − 1] = 𝑥[𝑛] − 𝑥[𝑛 − 4] fark denklemiyle ifade edilen DZD bir sistem için aşağıda verilen soruları cevaplayınız.<br>
Bu sistemin girişine 𝑥[𝑛] = 𝑢[𝑛 − 3] işaretini uygulayın. Bunun için önce 𝑥[𝑛]′i
n=0,1,...10 aralığı için üretip Scipy.signal kütüphanesinin convolve() fonksiyonuna ve aynı kütüphanenin lfilter() fonksiyonuna uygulayıp çıkış işaretini elde edip çizdirin (indislere dikkat edilmelidir). convolve() fonksiyonunu kullanırken üçüncü parametre olarak ‘same’ girildiğinde giriş ve çıkış işaretlerinin boyları aynı olur. Bu iki fonksiyon farklı sonuç üretti ise sebebini açıklayınız.

```python
import numpy as np
import scipy.signal as signal

def u(n):
    return np.where(n >= 0, 1, 0)
def delta(n):
    return np.where(n == 0, 1, 0)

n = np.arange(-20,21)
xn = u(n-3)
#𝑦[𝑛] − 0.8𝑦[𝑛 − 1] = 𝑥[𝑛] − 𝑥[𝑛 − 4]
b = [1, 0, 0, 0,-1]
a = [1, -0.8, 0, 0, 0]

ynlfilter = signal.lfilter(b, a, xn)

hn_analitic = ((0.8)**n)*u(n) - ((0.8)**(n-4))*u(n-4)
ynconv = signal.convolve(xn, hn_analitic, mode = "same")

#plot xn , hn , ynlfilter and ynconv with subplot
plt.subplot(2, 1, 1)
plt.stem(n, xn)
plt.title('x[n]')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.subplots_adjust(hspace=0.8)
plt.grid()

plt.subplot(2, 1, 2)
plt.stem(n, hn_analitic)
plt.title('h[n]')
plt.xlabel('n')
plt.ylabel('h[n]')
plt.subplots_adjust(hspace=0.8)
plt.grid()
plt.show()


plt.subplot(2, 1, 1)
plt.stem(n,ynlfilter)
plt.title('y[n] using lfilter')
plt.xlabel('n')
plt.ylabel('y[n]')
plt.grid()
plt.subplots_adjust(hspace=0.8)

plt.subplot(2, 1, 2)
plt.stem(n,ynconv)
plt.title('y[n] using conv')
plt.xlabel('n')
plt.ylabel('y[n]')
plt.grid()
plt.show()

```

Yukarda görüldüğü gibi lfilter ve conv ile üretlen y[n] çıkışları aynı olarak bulunmuştur.

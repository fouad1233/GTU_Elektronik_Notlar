## 2.1 Ayrık - zamanlı Fourier Dönüşümü (DTFT)

```python
w_cont= np.arange(0, 2*np.pi, np.pi/1000) 
H_abs=np.abs(2*np.cos(w_cont/2))
H_phase=np.array([-w/2 if w<np.pi else -w/2+np.pi for w in w_cont]) 
plt.subplot(121)
plt.plot(w_cont/np.pi,H_abs)
plt.title('$|X(e^{j\omega})|$')
plt.xlabel('Normalize frekans (x$\pi$)')
plt.ylabel('Genlik')
plt.subplots_adjust(wspace=0.5, hspace=0.5)
plt.subplot(122)
plt.plot(w_cont/np.pi,H_phase/np.pi)
plt.title('$ \measuredangle X(e^{j\omega})$')
plt.xlabel('Normalize frekans (x$\pi$)')
plt.ylabel('Faz (x$\pi$)')
plt.show()
```

## 2.2 Ayrık Fourier dönüşümü (DFT)

```python
def dirac(n):
    if n==0:
        return 1
    else:
        return 0
```

```python
N=8 
X_k=np.zeros([N],dtype=complex) 
for k in range(N):
    for n in range(N):
        X_k[k]=X_k[k]+(dirac(n)+dirac(n-1))*np.exp(-1j*(2*np.pi/N)*k*n)
        
X_k_abs=np.abs(X_k)
X_k_phase=np.angle(X_k)

plt.figure()
plt.subplot(121) 
plt.plot(w_cont/np.pi,H_abs,label='$|X(e^{j\omega})|$') 
w_discrete=np.arange(0,N)*(2*np.pi/N) 
plt.stem(w_discrete/np.pi,X_k_abs,'r-',label='$|X[k]|$',markerfmt='ro') 
plt.xlabel('$\omega$ (x$\pi$)')
plt.legend()

plt.subplot(122)
plt.plot(w_cont/np.pi,H_phase,label='$ \measuredangle X(e^{j\omega})$') 
plt.stem(w_discrete/np.pi,X_k_phase,'r-',label='$ \measuredangle X[k]$',markerfmt='ro')
plt.xlabel('$\omega$ (x$\pi$)')
plt.legend()
plt.show()
```

### 2.2.1 Frekansta örnekleme →Zamanda periyodiklik ilişkisi

Bir önceki bölümde 𝑁 noktalı DFT hesaplarken 𝑋[𝑘]’nın 𝑁 ile periyodik olduğundan bahsetmiştik. Sentez denklemine dikkat edilecek olursa, 𝑋[𝑘 + 𝑁] = 𝑋[𝑘] olduğundan, 𝑥[𝑛 + 𝑁]’in de 𝑥[𝑛]’e eşit olduğu görülecektir. Dolayısıyla, DFT ifadesinde 𝑥[𝑛]’in 𝑛 = 0,1, ... , 𝑁 − 1 aralığındaki değerlerinin her 𝑁noktada bir tekrarlandığı varsayımı mevcuttur. Dolayısıyla, 𝑁 noktalı DFT hesaplarken
işaretinizin 𝑁 ile periyodik olduğunu varsayarak DFT hesaplamanız gerekir. Aşağıda cos( (3 $\pi$/5) * N)
işaretini 10 noktalı ve 15 noktalı iki ayrı versiyonunu oluşturup Python’da DFT’lerini fft() ile hesaplayıp sadece genliklerini çizdirelim:

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.fftpack import fft , ifft
#10 uzunluklu versiyon
n1=np.arange(0,10)
x1=np.cos(3*np.pi/5*n1)
X1_abs=np.abs(fft(x1))
w_disc_1=n1*2*np.pi/len(n1) #0-2pi arası 2pi/10 adımlı vektör (2pinoktası dahil değil)
plt.figure()
plt.subplot(121)
plt.stem(w_disc_1/np.pi,X1_abs)
plt.xlabel('$\omega (x\pi)$') 
plt.title('$cos(\\frac{3\\pi}{5}n)$ 10 uzunluklu DFT')
#15 uzunluklu versiyon
n2=np.arange(0,15)
x2=np.cos(3*np.pi/5*n2)
X2_abs=np.abs(fft(x2))
w_disc_2=n2*2*np.pi/len(n2) #0-2pi arası 2pi/15 adımlı vektör (2pi noktası dahil değil)
plt.subplot(122)
plt.stem(w_disc_2/np.pi,X2_abs)
plt.xlabel('$\omega (x\pi)$') 
plt.title('$cos(\\frac{3\\pi}{5}n)$ 15 uzunluklu DFT')
plt.show()
```

$f f t(x)$ komutu ikinci bir parametre girilmediğinde DFT hesaplarken nokta sayısını $x^{\text {'}}$ in boyuna eşit seçer. Dolayısıyla, $x[n]=\cos \left(\frac{3 \pi}{5} n\right) n=0,1, \ldots, 9$ şeklinde 10 nokta için hesaplayıp $f f t$ () fonksiyonuna verdiğimizde $x^{\prime}$ i 10 noktalı periyodik bir işaret varsayıp 10 noktalı DFT hesaplarken, $n=0,1, \ldots, 14$ şeklinde 15 nokta için hesaplayıp fft() fonksiyonuna verdiğimizde $x$ 'i 15 noktalı periyodik bir işaret olarak varsayıp 15 noktalı DFT hesaplayacaktır. Yukarıda solda verilen grafik 10 noktalı DFT'yi, sağındaki ise bu işaretin 15 noktalı DFT'y' göstermektedir. $x[n]=\cos \left(\frac{3 \pi}{5} n\right)$ işaretinin Fourier dönüşümü alındığında solda gösterilen grafikte olduğu gibi $\omega=\frac{3 \pi}{5}$ ve $\omega=-\frac{3 \pi}{5}$ noktalarında dürtüler görmeyi bekleriz. Dikkat edilirse, $[0-2 \pi)$ arasını incelediğimiz için $-\frac{3 \pi}{5}=-\frac{3 \pi}{5}+2 \pi=\frac{7 \pi}{5}$ noktasına denk gelir. Yukarıda sağdaki grafikte ise ikiden fazla dürtü elde edilmektedir. Bunun sebebi, sağdaki grafik periyodu 10 olan $\cos \left(\frac{3 \pi}{5} n\right)$ grafiğinin değil, bir periyodu $x[n]=\cos \left(\frac{3 \pi}{5} n\right), n=$ $0,1,2, \ldots, 9,10,11,12,13,14$ olan ve her 15 noktada tekrar eden $x[n]$ işaretinin DFT'sini göstermektedir. Bu işaret aşağıdaki şekilde gösterilmektedir. İşaretin birinci ve üçüncü periyodu kolay görülebilmesi için farklı renkle boyanmıştır. Bu işaret, $\boldsymbol{\operatorname { c o s } (}\left(\frac{3 \pi}{5} n\right)$ işaretinin üç periyoduna eşit değildir, farklı bir işarettir ve dolayısıyla Fourier dönüşümü hesaplandığında yukarıda solda gösterildiği gibi $\omega=$ $\frac{3 \pi}{5}$ ve $\omega=-\frac{3 \pi}{5}$ noktalarında dürtüler olmayacaktır. Bu işaretin Fourier dönüşümü yukarıda sağdaki grafikte verildiği gibidir.

Özetle, $\cos \left(\frac{3 \pi}{5} n\right)$ işaretinin periyodu 10 örnek olduğu için $x$ vekrötünü 10 'un tam katları olacak şekilde oluşturduğunuzda DFT genlik grafiğinde sadece 2 tane dürtü görürsünüz. Aksi halde, hesapladığınız DFT $\cos \left(\frac{3 \pi}{5} n\right)$ işaretine ait değil, başka bir periyodik işarete ait olacaktır. Ödevlerde "tam 2 katını, tam 3 katını elde ediniz, vb" ifadelerin gerekçesi budur.

### 2.2.2 DFT genlik ve faz grafiklerinden işaret sentezi

Zamandaki kapalı formunu bilmediğiniz bir diziyi farklı frekanslardaki sinüs ve kosinüslerin doğrusal kombinasyonu şeklinde, dizinin DFT'sine bakarak yazabilirsiniz. Bunun için aşağıda verilen iki denklemi bilmeniz gerekmektedir.
$$
\begin{gathered}
\left.\tilde{x}[n]=\sum_{\substack{k=0}}^{N-1} a_k e^{+j \frac{2 \pi}{N} k n} \text { (DFS sentez denklemi) }\right ) \\
x[n]=\frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{+j \frac{2 \pi}{N} k n} \text { (DFT sentez denklemi) }
\end{gathered}
$$

Yukarıdaki iki denkleme baktığınızda periyodik olan ayrık zamanlı işaretin Fourier serisi katsayıları $a_k$ ile DFT katsayıları $X[k]$ arasında $a_k=\frac{X[k]}{N}$ ilişkisi olduğunu görüyoruz. Diğer bir değişle, DFT hesapladıktan sonra elde ettiğiniz diziyi dizinin boyuna bölerseniz elde ettiğiniz yeni dizi DFS katsayıları olur. Örneğin, $x=[0,2,0,-2]$ dizisi $\operatorname{Acos}\left(\omega_0 n+\phi\right)$ formatında bir sinüzoidal işaretin bir periyotta aldığı değerlere karşılık gelmektedir. Aşağıdaki kod, yukarıdaki dizinin fft() komutu ile DFT'sinin alınıp daha sonra işaretin uzunluğu olan $N=4$ 'e bölünüp elde ettiğimiz yeni kompleks dizinin genlik ve faz grafiklerini çizmektedir.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.fftpack import fft , ifft
x=np.array([0,2,0,-2])
N=len(x)
n=np.arange(0,N)
w_disc=n*2*np.pi/N # 0-2pi arasında(2pi noktası dahil değil) 2pi/4 adımlı vektör
H_abs=np.abs(fft(x)/N)
H_phase=np.angle(fft(x)/N)
plt.subplot(121)
plt.stem(w_disc/np.pi,H_abs)
plt.xlabel('$\omega (\\times \pi)$')
plt.title('$\\frac{1}{N}|X(e^{j\omega})|=|a_{k}|$') 
plt.subplot(122)
plt.stem(w_disc/np.pi,H_phase/np.pi)
plt.xlabel('$\omega (\\times \pi)$') 
plt.title('$\\frac{1}{N}\measuredangle X(e^{j\omega})(\\times \pi)=\measuredangle a_k(\\times \pi)$')
plt.show()
```

Yukarıda, solda $f f t($ ) fonksiyonu ile işaretin DFT'sini hesaplayıp işaretin uzunluğuna bölünmüş yeni dizinin genlik grafiği, sağda ise aynı dizinin faz grafiği verilmiştir. Genlik grafiğinden sadece birinci indisteki $0.5 \pi$ ve üçüncü indisteki $1.5 \pi$ frekanslarında DFS katsayılarının sıfırdan farklı değere eşit olduğunu görüyoruz. Bu nedenle DFS sentez denklemini kullanarak $\tilde{x}[n]$ 'i sentezlemek istersek sadece $a_1 e^{j 0.5 \pi n}$ ve $a_3 e^{j 1.5 \pi n}$ kompleks üstel işaretlerini toplamamız gerekir. $a_1$ katsayısını bulmak için:
$$
a_1=\left|a_1\right| e^{j \angle a_1}=1 e^{-j \frac{\pi}{2}}
$$

Yukarıda $\left|a_1\right|$ değerini genlik grafiğinde birinci indisteki değerden, $\angle a_1$ değerini ise faz grafiğindeki birinci indisteki değerden okuyoruz. Faz grafiğini çizdirirken okuma kolaylığı olması için vektör $\pi$ 'ye bölünmüştü. Bu nedenle, dik eksendeki değeri alırken $\pi$ ile çarpılması gerekir. Benzer şekilde $a_3{ }^{\prime}$ ü hesaplarsak:
$$
a_3=\left|a_3\right| e^{j \angle a_3}=1 e^{+j \frac{\pi}{2}}
$$

Sonuç olarak $\tilde{x}[n]$ aşağıdaki ifadeye eşit olur;
$$
\begin{aligned}
\tilde{x}[n] & =\sum_{k=0}^{N-1} a_k e^{+j \frac{2 \pi}{N} k n}=a_1 e^{j 0.5 \pi n}+a_3 e^{j 1.5 \pi n} \\
& =1 e^{-j \frac{\pi}{2}} e^{j 0.5 \pi n}+1 e^{+j \frac{\pi}{2}} e^{j 1.5 \pi}
\end{aligned}
$$
$e^{j 1.5 \pi n}=e^{j(2 \pi-0.5 \pi) n}=e^{j 2 \pi n} e^{-j 0.5 \pi n}=e^{-j 0.5 \pi n}$ olduğu için;
$$
\begin{aligned}
& =1 e^{-j \frac{\pi}{2}} e^{j 0.5 \pi n}+1 e^{+j \frac{\pi}{2}} e^{-j 0.5 \pi n} \\
& =e^{j\left(0.5 \pi n-\frac{\pi}{2}\right)}+e^{-j\left(0.5 \pi n-\frac{\pi}{2}\right)} \\
& =2 \cos \left(0.5 \pi n-\frac{\pi}{2}\right)
\end{aligned}
$$

Buradan $A=2, \omega_0=\frac{\pi}{2}$ ve $\phi=-\frac{\pi}{2}$ olduğunu görürüz.
Eğer $n=[0,1,2,3]$ vektörü için $2 \cos \left(0.5 \pi n-\frac{\pi}{2}\right)$ işaretini hesaplarsanız en başta verilen $[0,2,0,-2]$ dizisini elde edersiniz.

1) DZD olan bir sistemin dürtü cevab $h[n]=\delta[n]-\delta[n-1]$ olarak verilmektedir.<br>
a) $H\left(e^{j \omega}\right)$ 'yı elinizle hesaplayın ve yorum olarak ekleyin. Hesapladığınız bu fonksiyonunun genliği $\left|H\left(e^{j \omega}\right)\right|$ ile fazı $\angle H\left(e^{j \omega}\right)$ 'yı $\omega$ 'yı $0-2 \pi$ aralığında 1000 noktadan oluşacak şekilde oluşturup çizdirin.<br>
b) $\left|H\left(e^{j \omega}\right)\right|$ 'ya bakarak bu filtrenin nasıl bir karakteristiğe sahip olduğunu yorumlayın.<br>
c) $h[n]$ işaretini $n=0, \ldots, 15$ indislerinde tanıml1 16 noktalı olacak şekilde oluşturun. Oluşturduğunuz bu dizinin $f f t($ ) fonksiyonu ile 16 noktalı DFT'sini hesaplayıp genlik ve faz grafiklerini çizdirin.

## 1 a)

$\begin{aligned} & h[n]=\delta[n]-\delta[n-1] \\
& \text { ve } \\
& S[n] \Longleftrightarrow 1 \\
& S[n-1] \Longleftrightarrow e^{-j \omega} \\
& \Rightarrow H\left(e^{j \omega}\right)=1-e^{-j \omega} \\
& =\frac{2\left(1-e^{-j \omega}\right)}{2} \\
& =\frac{2 j e^{-j \frac{\omega}{2}}\left(e^{\frac{j \omega}{2}}-e^{-\frac{j \omega}{2}}\right)}{2 j} \\
& =2 j e^{-j \omega / 2} \sin \left(\frac{\omega}{2}\right) \\ & \text { ve } j=e^{-j \frac{3 \pi}{2}} \\ & \Rightarrow H\left(e^{j \omega}\right)=2 e^{-j\left(\omega / 2+\frac{3 {\pi}}{2}\right)} \sin \left(\frac{w}{2}\right) \\ & \left|H\left(e^{j \omega}\right)\right|=\left|2 \sin \left(\frac{w}{2}\right)\right| \\
& \measuredangle H\left(e^{j \omega}\right)=\left\{\begin{array}{l}-\frac{\omega}{2}-\frac{3 \pi}{2} \text { eğer } 0 \leqslant w< 2\pi  \\   \end{array}\right. \\ & \end{aligned}$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import fft , ifft
N=16
n=np.arange(0,N)
#$h[n]=\delta[n]-\delta[n-1]$
def dirac(n):
    return np.where(n==0,1,0)

h_n=dirac(n)-dirac(n-1)
H_k=fft(h_n,N)
H_k_abs=np.abs(H_k)
H_k_phase=np.angle(H_k)
w_discrete=np.arange(0,N)*(2*np.pi/N)
plt.figure()
plt.subplot(121)
plt.stem(w_discrete/np.pi,H_k_abs)
plt.xlabel('$\omega (x\pi)$')
plt.ylabel('$|H[k]|$')
plt.title('$|H[k]|$')
plt.subplots_adjust(wspace=0.5, hspace=0.5)
plt.subplot(122)
plt.stem(w_discrete/np.pi,H_k_phase)
plt.xlabel('$\omega (x\pi)$')
plt.ylabel('Faz (x$\pi$)')
plt.title('$ \measuredangle H[k]$')
plt.show()

```

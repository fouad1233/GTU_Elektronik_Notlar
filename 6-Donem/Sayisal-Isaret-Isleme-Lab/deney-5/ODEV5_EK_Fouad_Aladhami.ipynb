## SORU-3: Soru-1’i 𝐹𝑠 = 1000⁡𝐻𝑧 için tekrarlayınız.


𝑥1(𝑡) = 7cos⁡(200𝜋𝑡) ve 𝑥2(𝑡) = 3 cos(200𝜋𝑡) + 4cos⁡(1800𝜋𝑡) işaretlerini 𝐹𝑠 = 2200⁡𝐻𝑧 ile örnekleyiniz.
### a)
Örnekleme sonucunda elde ettiğiniz 𝑥1[𝑛] ve 𝑥2[𝑛] sinyallerinin grafiklerini iki periyot için çizdirerek karşılaştırınız. İki işaret arasında her hangi bir fark var mı? Neden?

```python
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi 
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
F=100 # x işaretinin temel frekansı
T=1/F # x işaretinin temel periyodu
t=np.arange(0.,2*T,0.0001) # x işaretine ait t değişkeninin tanımlanması(2 periyot boyunca)
x1=7*np.cos(200*np.pi*t) # x(t) işaretinin tanımlanması
x2=3*np.cos(200*np.pi*t) + 4*np.cos(1800*np.pi*t) # x(t) işaretinin tanımlanması

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.xlabel("t") # grafiğin x ekseninin isimlendirilmesi 
plt.title("x1(t)=7*cos(200*pi*t)") # Grafik başlığının oluşturulması
plt.plot(t,x1) # işaretin grafiğinin çizdirilmesi
plt.subplot(1, 2, 2)
plt.xlabel("t") # grafiğin x ekseninin isimlendirilmesi 
plt.title("x2(t)=3*cos(200*pi*t) + 4*cos(1800*pi*t)") # Grafik başlığının oluşturulması
plt.plot(t,x2) # işaretin grafiğinin çizdirilmesi
plt.show() # grafiklerin gösterilmesini sağlar

Fs=1000
Ts=1/Fs
N=10
# örnekleme frekansının tanımlanması # örnekleme periyodunun tanımlanması
# örnek sayısının tanımlanması
n=np.arange(0,2*N) # örnekleme indisinin 0'dan iki periyot olacak şekilde array olarak tanımlanması
x1n=5*np.cos(200*np.pi*n*Ts) #örneklenmiş x[n] işaretinin tanımlanması 
x2n = 3*np.cos(200*np.pi*n*Ts) + 4*np.cos(1800*np.pi*n*Ts) 
# 2pik = 1800piN0 => N0 = k/900
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.stem(n,x1n) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x1[n]") # x[n] işaretinin grafiğinin isimlendirilmesi 
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi 
plt.ylabel("x1[n]") # grafiğin y ekseninin isimlendirilmesi
plt.subplot(1, 2, 2)
plt.stem(n,x2n) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x2[n]") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x2[n]") # grafiğin y ekseninin isimlendirilmesi

plt.show() # grafiğin gösterilmesi
```

İki işaretin örneklenmiş halleri birbiri ile aynıdır.<br>

$x2$ işaretinin frekansı $f_{N2} = 900 Hz $ <br>

Örnekleme freaknsı $ F_s = 1000 hz !> 900 * 2 >  100 * 2 $<br>

```python
###𝑥1[𝑛] ve 𝑥2[𝑛] sinyallerinin Fourier transformlarını elde ederek işaretleri frekans domaininde inceleyiniz. Elde ettiğiniz 𝑋1(𝜔) ve 𝑋2(𝜔) grafiklerini beklediğiniz gibi mi? Grafikleri karşılaştırarak grafikler arasında fark olup olmadığını açıklayarak ve nedenini irdeleyiniz.

w = np.arange(-np.pi, np.pi, 2*np.pi/N)
x1_w = np.fft.fftshift(np.fft.fft(x1n, N)/N)
x2_w = np.fft.fftshift(np.fft.fft(x2n, N)/N)
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.stem(w/np.pi, np.abs(x1_w))
plt.title("X1(w)'nin genlik grafiği")
plt.xlabel('w(x pi)')
plt.ylabel('|X1(w)|')
plt.grid()
plt.subplot(1, 2, 2)
plt.stem(w/np.pi, np.abs(x2_w))
plt.title("X2(w)'nin genlik grafiği")
plt.xlabel('w(x pi)')
plt.ylabel('|X2(w)|')
plt.grid()
plt.show()

```

```python
"""Ekte verilen odev.wav dosyasını okuyup örnekleme frekansını belirleyiniz. İşlemlerin devamı için ses işaretinin sadece tek kanalını kullanınız. İşareti dinleyiniz ve aşağıdaki işlemlerden sonra elde edeceğiniz ses işaretini dinleyerek karşılaştırıp yorumlayınız:<br>
a) Ses işaretini önce 𝑀 = 2 ile seyrek örnekleme, elde edilen yeni işarete ise 𝐿 = 2 ile sık örneklemeye
tabi tutunuz. Elde ettiğiniz nihai ses işaretini dinleyiniz ve orijinal ses işareti ile karşılaştırıp sebeplerini açıklayınız. Açıklamanızı gerektiğinde işaretin Fourier dönüşümünü elde edip inceleyerek de yapabilirsiniz. Sık örnekleme için interpolasyon işlemini lineer enterpolasyon ve ideal AGF kullanıp karşılaştırınız."""

import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
from scipy.io.wavfile import read # wav dosyasını okumak için gerekli kütüphanenin aktifleştirilmesi

Fs, x = read('ELM368 Ödev-5 odev.wav') # ses dosyasının okunması

print(x.shape) # ses dosyasının boyutunun yazdırılması
print(Fs) # ses dosyasının örnekleme frekansının yazdırılması
```

```python
xn = x[:,1] # ses dosyasının birinci kanalının alınması
N = len(xn) # ses dosyasının uzunluğunun alınması
n = np.arange(0, N) # ses dosyasının indislerinin alınması
t = n/Fs # ses dosyasının zamanının hesaplanması
plt.figure(figsize=(10, 4))
plt.stem(n, xn) # ses dosyasının grafiğinin çizdirilmesi
plt.title("x[n]") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.show() # grafiğin gösterilmesi
```

```python
xn = x[:,1] # ses dosyasının birinci kanalının alınması
N = len(xn) # ses dosyasının uzunluğunun alınması
n = np.arange(0, N) # ses dosyasının indislerinin alınması
t = n/Fs # ses dosyasının zamanının hesaplanması
plt.figure(figsize=(10, 4))
plt.stem(n, xn) # ses dosyasının grafiğinin çizdirilmesi
plt.title("x[n]") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.show() # grafiğin gösterilmesi
```

```python
M = 6 # Downsample faktörü
xn_d = xn[::M] # Downsample işlemi
N_d = len(xn_d) # Downsample sonucu oluşan dizinin uzunluğu
n_d = np.arange(0, N_d) # Downsample sonucu oluşan dizinin indisleri
t_d = n_d/(Fs/M) # Downsample sonucu oluşan dizinin zamanının hesaplanması

plt.figure(figsize=(10, 4))
plt.stem(n_d, xn_d) # ses dosyasının grafiğinin çizdirilmesi
plt.title("x[n] Downsample") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
```

```python
L = 6 # upsample faktörü
N_u = N_d*L # upsample sonucu oluşan dizinin uzunluğu
xn_u = np.zeros(N_u) # upsample sonucu oluşan dizinin tanımlanması
xn_u[::L] = xn_d # upsample işlemi
n_u = np.arange(0, N_u) # upsample sonucu oluşan dizinin indislerinin tanımlanması

#interpolasyon işlemi 
# hn işaretinin tanımlanması L = 6 için for loop ile
hn = [abs(i)/L for i in range(1, L+1)] + [abs(i)/L for i in range(L-1, 0, -1)]
print(hn)

xn_i = np.convolve(xn_u, hn, 'same') # interpolasyon işlemi
n_i = np.arange(0, len(xn_i)) # interpolasyon işlemi sonucu oluşan dizinin indislerinin tanımlanması
plt.figure(figsize=(10, 4))
plt.stem(n_u, xn_i) # ses dosyasının grafiğinin çizdirilmesi
plt.title("x_i[n] upsample") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_i[n]") # grafiğin y ekseninin isimlendirilmesi

```

Bu durumda verilen 𝑥[𝑛] işareti için Sistem1 çıkışı olan 𝑦1[𝑛] işaretini ve Sistem2 çıkışı olan 𝑦2[𝑛] işaretini, gerekli Python kodunu yazarak elde ediniz ve elde ettiğiniz 𝑦1[𝑛] ve 𝑦2[𝑛] işaretlerinin grafiklerini çizdirerek karşılaştırınız ve sonuçları yorumlayınız.

```python
"""
Ayrık zamanlı 𝑥[𝑛]=[2⁡,3.2⁡,4⁡,−5,−6.7⁡,7.8⁡,15]⁡işareti aşağıda verilen Sistem1 ve Sistem2 sistemlerine ayrı ayrı girdi olarak verilmektedir. Bu durumda verilen 𝑥[𝑛] işareti için Sistem1 çıkışı olan 𝑦1[𝑛] işaretini ve Sistem2 çıkışı olan 𝑦2[𝑛] işaretini, gerekli Python kodunu yazarak elde ediniz ve elde ettiğiniz 𝑦1[𝑛] ve 𝑦2[𝑛] işaretlerinin grafiklerini çizdirerek karşılaştırınız ve sonuçları yorumlayınız.

"""
import numpy as np # dizi işlemleri için gerekli kütüphanenin aktifleştirilmesi
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi

xn = [2, 3.2, 4, -5, -6.7, 7.8, 15] # x[n] işaretinin tanımlanması
n = np.arange(0, len(xn)) # x[n] işaretinin indislerinin tanımlanması
L1 = 2 # Sistem1 için L1 değerinin tanımlanması
L2 = 3 # Sistem1 için L2 değerinin tanımlanması
x1_L1 = np.zeros(len(xn)*L1) # Sistem1 için x1[n] işaretinin tanımlanması
x1_L1[::L1] = xn # Sistem1 için x1[n] işaretinin tanımlanması
hL1 = [abs(i)/L1 for i in range(1, L1+1)] + [abs(i)/L1 for i in range(L1-1, 0, -1)]
x1_L1_hL1 = np.convolve(x1_L1, hL1, 'same') # Sistem1 için x1[n]*h1[n] işaretinin tanımlanması
x1_L2 = np.zeros(len(x1_L1_hL1)*L2) # Sistem1 için x1_L2[n] işaretinin tanımlanması
x1_L2[::L2] = x1_L1_hL1 # Sistem1 için x1_L2[n] işaretinin tanımlanması
hL2 = [abs(i)/L2 for i in range(1, L2+1)] + [abs(i)/L2 for i in range(L2-1, 0, -1)]
y1 = np.convolve(x1_L2, hL2, 'same') # Sistem1 için y1[n] işaretinin tanımlanması
plt.figure(figsize=(8, 12))
plt.subplot(5, 1, 1)
# tüm grafiklerin çizdirilmesi
plt.stem(n, xn, label="x[n]") # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.subplot(5, 1, 2)
plt.stem( x1_L1, label="x1[n]") # x1[n] işaretinin grafiğinin çizdirilmesi
plt.title("x1[n]") # x1[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x1[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.subplot(5, 1, 3)
plt.stem(x1_L1_hL1, label="x1[n]*h1[n]") # x1[n]*h1[n] işaretinin grafiğinin çizdirilmesi
plt.title("x1[n]*h1[n]") # x1[n]*h1[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x1[n]*h1[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.subplot(5, 1, 4)
plt.stem(x1_L2, label="x1_L2[n]") # x1_L2[n] işaretinin grafiğinin çizdirilmesi
plt.title("x1_L2[n]") # x1_L2[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x1_L2[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.subplot(5, 1, 5)
plt.stem(y1, label="y1[n]") # y1[n] işaretinin grafiğinin çizdirilmesi
plt.title("y1[n]") # y1[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("n") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("y1[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.tight_layout()
plt.show()


```

## 3.1 Sürekli zamanlı bir işaretin örneklenerek ayrık zamanlı işaretin elde edilmesi ve örnekleme frekansının etkisinin incelenmesi

Sürekli zamanlı bir 𝑥(𝑡) sinyali aşağıda verilen blok şemadaki gibi örneklenmektedir. 𝑥(𝑡) sinyalinden 𝑥[𝑛] sinyalinin teorik olarak elde edilmesi ise Eş.(1)’deki gibi olmaktadır.
$$ 𝑥[𝑛] = 𝑥(𝑛𝑇_s ) = 𝑥(𝑡)|_{(t<- nT_s)}  (1) $$

Yukarıdaki eşitlikte verildiği üzere ayrık zamanlı 𝑥[𝑛] işaretini elde etmek için sürekli zamanlı 𝑥(𝑡) sinyalinde 𝑡 değişkeni yerine $ 𝑛𝑇_s $ yazılmaktadır.Eşitlikte yer alan $ 𝑇_s $ , örnekleme periyodunu belirtmektedir ve örnekleme frekansı olan $ F_s $ ile aralarında şu bağıntı vardır:
$$ 𝑇_s  = 1/F_s$$

Sürekli zamanlı bir 𝑥(𝑡) sinyali örneklenirken aşağıda verilen Nyquist şartının sağlanıp sağlanmadığına dikkat edilmesi gerekir. Aksi halde birbirini 2𝜋’ye tamamlayan ayrık zamanlı sinyallerde örtüşme meydana gelir. $$ F_s \geq 2F_N $$

Yukarıda verilen $𝐹_N $ frekansı, 𝑥(𝑡) işaretinin en yüksek frekans bileşenini temsil etmektedir.

Örnek 2: Örnek 1’de verilen 𝑥(𝑡) = 5cos⁡(200𝜋𝑡) işaretini $𝐹_s = 2200⁡𝐻𝑧$ için örnekleyerek 
grafiğini iki periyot boyunca çizdiriniz.<br> 

𝑥(𝑡) = 5cos⁡(200𝜋𝑡) işaretinin frekansı $𝐹_N = 100⁡𝐻𝑧$ ’ dir. $𝐹_s = 2200⁡𝐻𝑧$ için Nyquist şartına bakıldığında şartın sağlandığı açıkça görülmektedir.

Aşağıda 𝑥[𝑛] ayrık zamanlı işaretinin elde edilmesi ve iki periyot boyunca çizdirilmesi için yazılan kodlar verilmektedir.⁡𝑥[𝑛] işareti oluşturulurken bir periyottaki toplam örnek sayısını temsil eden 𝑁 değerinin belirlenmesi gerekmektedir. Bu değerin belirlenmesi için izlenen yol şu şekildedir:


1- 𝑥[𝑛] işareti Eş.(1) yardımıyla elde edilir. <br>
2- 𝑥[𝑛] işaretinin 𝜔 açısal frekans değeri okunur.<br>
3- 𝜔 = 2𝜋𝑘/N eşitliğini sağlayan 𝑁 değeri bulunur.<br>

Bu işlemleri sırası ile yaparsak: <br>
1-⁡𝑥[𝑛] = 5cos(𝜋𝑛/11)<br>
2- w = 𝜋 /11<br>
3- 𝜔 = 𝜋/11 = 2𝜋𝑘/N eşitliğinden 𝑘 = 1 için 𝑁 değeri 22 örnek olarak bulunur.<br>

```python
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi 
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
Fs=2200
Ts=1/Fs
N=22
# örnekleme frekansının tanımlanması # örnekleme periyodunun tanımlanması
# örnek sayısının tanımlanması
n=np.arange(0.,2*N) # örnekleme indisinin 0'dan iki periyot olacak şekilde array olarak tanımlanması
xn=5*np.cos(200*np.pi*n*Ts) #örneklenmiş x[n] işaretinin tanımlanması 
plt.stem(n,xn) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi 
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi 
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.show() # grafiğin gösterilmesi
```

## 3.2 Örtüşmenin zaman domaininde incelenmesi

```python
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
t = np.arange(0, 2/60, 1/4000) # t değişkeninin tanımlanması
xt_1 = np.cos(2*np.pi*60*t + np.pi/3) # x1(t) işaretinin tanımlanması
xt_2 = np.cos(2*np.pi*340*t - np.pi/3) # x2(t) işaretinin tanımlanması
xt_3 = np.cos(2*np.pi*460*t + np.pi/3) # x3(t) işaretinin tanımlanması

Fs = 400 # örnekleme frekansının tanımlanması
Ts = 1/Fs # örnekleme periyodunun tanımlanması
nTs = np.arange(0, 2/60, Ts) # nTs değişkeninin tanımlanması

xnTs_1 = np.cos(2*np.pi*60*nTs + np.pi/3) # x1(nTs) işaretinin tanımlanması
xnTs_2 = np.cos(2*np.pi*340*nTs - np.pi/3) # x2(nTs) işaretinin tanımlanması
xnTs_3 = np.cos(2*np.pi*460*nTs + np.pi/3) # x3(nTs) işaretinin tanımlanması

plt.figure(figsize=(8, 8))
# x1(t) işaretinin grafiğinin çizdirilmesi
plt.subplot(3, 1, 1)

plt.title('x1(t) and x1(nTs)')
plt.xlabel('t')
plt.ylabel('x1(t)')
plt.plot(t, xt_1, color='orange')
plt.stem(nTs, xnTs_1)
plt.grid()

# x2(t) işaretinin grafiğinin çizdirilmesi
plt.subplot(3, 1, 2)
plt.title('x2(t) and x2(nTs)')
plt.xlabel('t')
plt.ylabel('x2(t)')
plt.plot(t, xt_2, color='orange')
plt.stem(nTs, xnTs_2)
plt.grid()

# x3(t) işaretinin grafiğinin çizdirilmesi
plt.subplot(3, 1, 3)
plt.title('x3(t) and x3(nTs)')
plt.xlabel('t')
plt.ylabel('x3(t)')
plt.plot(t, xt_3, color='orange')
plt.stem(nTs, xnTs_3)
plt.grid()
plt.tight_layout()


```

## 3.3 Örtüşmenin frekans domaininde incelenmesi

Örnek 4

```python


def signall(Fs,N,k):
    F = 100
    T = 1/F
    t = np.linspace(-k*T/2, k*T/2, 1000)
    xt = np.sinc(2*F*t)**2
    Ts = 1/Fs
    n = np.arange(-N/2, N/2)
    xn = np.sinc(2*F*n*Ts)**2
    
    #xn işaretinin fourier tranformu 
    w = np.arange(-np.pi, np.pi, 2*np.pi/N)
    xw = np.fft.fftshift(np.fft.fft(xn, N)/N)
    
    #plot xt , xn ve xw
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 3, 1)
    plt.title('x(t) = sinc^2(2*{}*t)'.format(F))
    plt.xlabel('t')
    plt.ylabel('x(t)')
    plt.plot(t, xt)
    plt.grid()
    
    plt.subplot(1, 3, 2)
    plt.title('x[n] = sinc^2(2*{}*n*Ts)'.format(F)+'when Ts = 1/{}'.format(Fs))
    plt.xlabel('n')
    plt.ylabel('x[n]')
    plt.stem(n, xn)
    plt.grid()
    
    plt.subplot(1, 3, 3)
    plt.title('X(w)')
    plt.xlabel('w')
    plt.ylabel('X(w)')
    plt.stem(w, np.abs(xw))
    plt.grid()
    plt.tight_layout()
    plt.subplots_adjust(wspace=.5)
    
signall(1000, 30, 3)
signall(600, 30, 5)
signall(300, 30, 10)
signall(100, 30, 30)

```

## 3.4 Ayrık zamanlı bir işaretin Fourier transformunun genlik ve faz grafiğinin çizdirilmesi

Örnek 5: Örnek 1’de verilen sürekli zamanlı 𝑥(𝑡) = 5cos⁡(200𝜋𝑡) işaretinin 𝐹s= 2200⁡𝐻𝑧 ile örneklenmesi sonucunda (Örnek 2’de) elde edilen ayrık zamanlı x[𝑛] işaretini, frekans domaininde inceleyelim. <br>
𝑥[𝑛] = 5 cos(𝜋𝑛/11) olmak üzere 𝑥[𝑛] işaretinin Fourier transformunun grafiği aşağıda verilmektedir. 

```python
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi 
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
Fs=2200
Ts=1/Fs
N=22
# örnekleme frekansının tanımlanması # örnekleme periyodunun tanımlanması
# örnek sayısının tanımlanması
n=np.arange(0.,2*N) # örnekleme indisinin 0'dan iki periyot olacak şekilde array olarak tanımlanması
xn=5*np.cos(200*np.pi*n*Ts) #örneklenmiş x[n] işaretinin tanımlanması 
plt.stem(n,xn) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi 
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi 
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.show() # grafiğin gösterilmesi

#Şimdi bu x[n] işaretinin fourier transformunu hesaplayalım
w = np.linspace(-np.pi, np.pi, N)
x_w = np.fft.fftshift(np.fft.fft(xn, N)/N)
plt.figure(figsize=(6, 4))
plt.stem(w/np.pi, np.abs(x_w))
plt.title("X(w)'nin genlik grafiği")
plt.xlabel('w(x pi)')
plt.ylabel('|X(w)|')
plt.grid()
plt.show()
plt.figure(figsize=(6, 4))
plt.stem(w/np.pi, np.angle(x_w))
plt.title("X(w)'nin faz grafiği")
plt.xlabel('w(x pi)')
plt.ylabel('$ \measuredangle X(w) $')
plt.grid()


```

# 4 Sürekli zamanlı işaretin yeniden elde edilmesi

## 4.1 Sürekli zamanlı sinyalin frekans domaininden elde edilmesi

```python
import matplotlib.pyplot as plt # grafik çizimi için gerekli kütüphanenin aktifleştirilmesi 
import numpy as np # dizi işlemleri için kütüphane aktif hale getirilir.
F=100 # x işaretinin temel frekansı
T=1/F # x işaretinin temel periyodu
k = 4
t=np.arange(0.,k*T,0.0001) # x işaretine ait t değişkeninin tanımlanması(2 periyot boyunca)
x=5*np.cos(200*np.pi*t) # x(t) işaretinin tanımlanması
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.xlabel("t") # grafiğin x ekseninin isimlendirilmesi 
plt.title("x(t)=5*cos(200*pi*t)") # Grafik başlığının oluşturulması
plt.plot(t,x) # işaretin grafiğinin çizdirilmesi

Fs=2200
Ts=1/Fs
N=22
# örnekleme frekansının tanımlanması # örnekleme periyodunun tanımlanması
# örnek sayısının tanımlanması
n=np.arange(0.,k*N) # örnekleme indisinin 0'dan iki periyot olacak şekilde array olarak tanımlanması
xn=5*np.cos(200*np.pi*n*Ts) #örneklenmiş x[n] işaretinin tanımlanması 
plt.subplot(1, 2, 2)
plt.stem(n,xn) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi 
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi 
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.show() # grafiğin gösterilmesi
```

```python
x_sinc = np.zeros([k*N, np.size(x , axis=0)])
plt.figure(figsize=(12, 4))
for ni in n:
    #print(ni)
    ni = int(ni)
    x_sinc[ni,:] = xn[ni]*np.sinc((t-ni*Ts)/Ts )
    plt.plot( x_sinc[ni,:])

plt.title('İşaretin zaman domaininde sinc(.) ile konvolüsyonu sonucunda elde edilen bileşenler')
plt.xlabel('örnek')
plt.ylabel('x[n] * sinc(t-nTs)')
```

```python
xr_t = np.sum(x_sinc, axis=0)
plt.figure(figsize=(8, 4))
plt.xlabel("t") # grafiğin x ekseninin isimlendirilmesi
plt.title("x_r(t)") # Grafik başlığının oluşturulması
plt.plot(t, xr_t) # işaretin grafiğinin çizdirilmesi
plt.show() # grafiklerin gösterilmesini sağlar
```

```python
import matplotlib.pyplot as plt # for graph plotting
import numpy as np 			# for array operations

# Sampling of the signal x(t) = 5*cos(200*pi*t) with a sampling frequency Fs
FS = 1200 					# sampling frequency
Ts = 1/FS 					# sampling period
N = 12 						# number of samples
n = np.arange(0, N) 			# array of indices for one period of sampling
xn = 5*np.cos(200*np.pi*n*Ts)  # sampled x[n] signal

plt.figure(figsize= (10, 10))
plt.subplot(3,1,1)
plt.stem(n, xn)
plt.xlabel("n (sample)")
plt.ylabel("x[n]")

L = 2 
Nn_u = N*L 					# upsampling factor
xn_u = np.zeros(Nn_u) 		# initialize upsampled signal array
xn_u[::L] = xn 				# insert x[n] values into xn_u array every L samples
n_u = np.arange(0, Nn_u) 		# indices for upsampled signal

plt.subplot(3,1,2)
plt.stem(n_u, xn_u)
plt.ylabel('$x_u[n]$')
plt.xlabel('n (sample)')

hn = np.array([0, 1/2, 1, 1/2, 0])  # linear interpolation h[n] for L=2
xn_i = np.convolve(xn_u, hn, 'full')  # convolution for interpolation
n_i = np.arange(0, len(xn_i)) 		# indices for interpolated signal

plt.subplot(3,1,3)
plt.stem(n_i, xn_i)
plt.xlabel("n (sample)")
plt.ylabel("$x_i[n]$")
plt.tight_layout()
plt.show()
```

Örnekleme frekansı 40 kHz olan 𝑥[𝑛] işaretinin örnekleme frekansının 50 kHz olması için yapılması gereken sık ve seyrek örnekleme oranlarını belirleyiniz ve aşağıdaki soruları belirlediğiniz değerleri kullanarak cevaplayınız.

## b)
Blok diyagramda yer alan interpolasyon işlemi için lineer interpolasyonu kullanınız. Bu amaçla lineer interpolasyon sisteminin h𝑙𝑖𝑛[𝑛] dürtü cevabını belirleyiniz. Belirlediğiniz değerleri kullanarak 𝑥𝑑[𝑛] işaretini elde etmek için gereken kodu jupyter platformunda yazınız. 𝑥𝑑[𝑛] işaretini elde ederken bulmanız gereken 𝑥𝑢[𝑛], 𝑥𝑖[𝑛] ve 𝑥𝑑[𝑛] işaretleriningrafiklerini de çizdirerek her aşamada neler olduğunu inceleyiniz. Ardından elde ettiğiniz 𝑥[𝑛], 𝑥𝑢[𝑛], 𝑥𝑖[𝑛] ve 𝑥𝑑[𝑛] işaretlerinin Fourier dönüşümlerinin genlik grafiklerini çizdirerek adım adım işlemlerin frekans domaindeki işaretlere olan etkilerini yorumlayınız.

```python
import numpy as np
import matplotlib.pyplot as plt

Fs = 40000 # örnekleme frekansı
Ts = 1/Fs # örnekleme periyodu
N = 20 # örnek sayısı
n = np.arange(0, N) # örnekleme indisi
xn = np.cos(np.pi*n/5) # x[n] işaretinin tanımlanması
plt.figure(figsize=(12, 12))
plt.subplot(5, 1, 1)
plt.stem(n, xn) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

## 40 *5 = 200 => 200/4 = 40 Khz
L = 5 # upsample faktörü
M = 4 # Downsample faktörü

#upsample işlemi
xn_u = np.zeros(N*L) # upsample işlemi sonucu oluşan dizinin tanımlanması
xn_u[::L] = xn # upsample işlemi
n_u = np.arange(0, N*L) # upsample işlemi sonucu oluşan dizinin indislerinin tanımlanması

#upsample sinyali çizdirme 
plt.subplot(5, 1, 2)
plt.stem(n_u, xn_u) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x_u[n] upsample") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_u[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

#interpolasyon işlemi
# h[n] işaretinin tanımlanması
hn = np.array([1/5, 2/5, 3/5, 4/5, 1, 4/5, 3/5, 2/5, 1/5]) # lineer interpolasyon h[n] işaretinin tanımlanması
nhn = np.arange(-4, 5, 1) # h[n] işaretinin indislerinin tanımlanması
# h[n] işaretinin grafiğinin çizdirilmesi
plt.subplot(5, 1, 3)
plt.stem(nhn, hn)
plt.title("h[n] lineer interpolasyon")
plt.xlabel("örnek")
plt.ylabel("h[n]")
plt.grid()

#interpolasyon işlemi
xn_i = np.convolve(xn_u, hn, 'same') # interpolasyon işlemi
n_i = np.arange(0, len(xn_i)) # interpolasyon işlemi sonucu oluşan dizinin indislerinin tanımlanması
N_i = len(xn_i) # interpolasyon işlemi sonucu oluşan dizinin uzunluğu
# interpolasyon işlemi sonucu oluşan dizinin grafiğinin çizdirilmesi
plt.subplot(5, 1, 4)
plt.stem(n_i, xn_i)
plt.title("x_i[n] interpolasyon")
plt.xlabel("örnek")
plt.ylabel("x_i[n]")
plt.grid()

#downsample işlemi
xn_d = xn_i[::M] # Downsample işlemi
N_d = len(xn_d) # Downsample sonucu oluşan dizinin uzunluğu
n_d = np.arange(0, N_d) # Downsample sonucu oluşan dizinin indisleri
# Downsample işlemi sonucu oluşan dizinin grafiğinin çizdirilmesi
plt.subplot(5, 1, 5)
plt.stem(n_d, xn_d)
plt.title("x_d[n] Downsample") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_d[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.tight_layout()

plt.show() # grafiğin gösterilmesi




```

```python
#xn işaretinin fourier transformu 
w = np.arange(-np.pi, np.pi, 2*np.pi/N)
xw = np.fft.fftshift(np.fft.fft(xn, N)/N)
plt.figure(figsize=(12, 8))
plt.subplot(3, 1, 1)
plt.stem(w/np.pi, np.abs(xw)) # x[n] işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("X(w)'nin genlik grafiği") # x[n] işaretinin fourier transformununun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|X(w)|") # grafiğin y ekseninin isimlendirilmesi
plt.grid()


#xi işaretinin fourier transformu
w_i = np.arange(-np.pi, np.pi, 2*np.pi/N_i)
xi_w = np.fft.fftshift(np.fft.fft(xn_i, N_i)/N_i)
plt.subplot(3, 1, 2)
plt.stem(w_i/np.pi, np.abs(xi_w)) # x[n] işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("X_i(w)'nin genlik grafiği") # x[n] işaretinin fourier transformunun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|X_i(w)|") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

#xd işaretinin fourier transformu
w_d = np.arange(-np.pi, np.pi, 2*np.pi/N_d)
xd_w = np.fft.fftshift(np.fft.fft(xn_d, N_d)/N_d)
plt.subplot(3, 1, 3)
plt.stem(w_d/np.pi, np.abs(xd_w)) # x[n] işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("X_d(w)'nin genlik grafiği") # x[n] işaretinin fourier transformununun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|X_d(w)|") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.tight_layout()
plt.show() # grafiğin gösterilmesi


```

### 2.c
Seçenek b’deki interpolasyon işlemi için Şekil 29’de yer alan ideal AGF filtresinin (𝐻(𝜔)) kesim frekansı ve kazanç parametrelerini belirleyerek b seçeneğinde yapılan işlemleri tekrarlayınız.

```python

import numpy as np
import matplotlib.pyplot as plt

Fs = 40000 # örnekleme frekansı
Ts = 1/Fs # örnekleme periyodu
N = 20 # örnek sayısı
n = np.arange(0, N) # örnekleme indisi
xn = np.cos(np.pi*n/5) # x[n] işaretinin tanımlanması
plt.figure(figsize=(12, 12))
plt.subplot(5, 1, 1)
plt.stem(n, xn) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x[n]") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

## 40 *5 = 200 => 200/4 = 40 Khz
L = 5 # upsample faktörü
M = 4 # Downsample faktörü

#upsample işlemi
xn_u = np.zeros(N*L) # upsample işlemi sonucu oluşan dizinin tanımlanması
xn_u[::L] = xn # upsample işlemi
n_u = np.arange(0, N*L) # upsample işlemi sonucu oluşan dizinin indislerinin tanımlanması

#upsample sinyali çizdirme 
plt.subplot(5, 1, 2)
plt.stem(n_u, xn_u) # x[n] işaretinin grafiğinin çizdirilmesi
plt.title("x_u[n] upsample") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_u[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

#interpolasyon işlemi fourier dönüşümü ile
xw_u = np.fft.fftshift(np.fft.fft(xn_u, N*L)/(N*L)) # upsample işaretinin fourier transformunun hesaplanması
w_u = np.arange(-np.pi, np.pi, 2*np.pi/(N*L)) # upsample işaretinin fourier transformunun frekansının hesaplanması
plt.subplot(5, 1, 3)
plt.stem(w_u/np.pi, np.abs(xw_u)) # upsample işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("X_u(w) upsample") # upsample işaretinin fourier transformunun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|X_u(w)|") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

#interpolasyon işlemi
# H(w) işaretinin tanımlanması
hw = np.zeros(N*L) # H(w) işaretinin tanımlanması
hw = np.where(np.abs(w_u) <= np.pi/5, L, 0) # H(w) işaretinin tanımlanması
xw_i = xw_u * hw # interpolasyon işlemi
plt.subplot(5, 1, 4)
plt.stem(w_u/np.pi, np.abs(hw)) # h[n] işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("H(w) upsample") # h[n] işaretinin fourier transformunun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|H(w)|") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
#filtre ile işaretin çarpılması
plt.subplot(5, 1, 5)
plt.stem(w_u/np.pi, np.abs(xw_i)) # x[n] işaretinin fourier transformunun genlik grafiğinin çizdirilmesi
plt.title("X_i(w) upsample") # x[n] işaretinin fourier transformunun genlik grafiğinin isimlendirilmesi
plt.xlabel("w(x pi)") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("|X_i(w)|") # gragiğin y ekseninin isimlendirilmesi
plt.grid()
plt.tight_layout()
plt.show() # grafiğin gösterilmesi

#ters fourier dönüşümü ile işaretin zaman domainine dönüşümü
xn_i = np.real(np.fft.ifft(np.fft.ifftshift(xw_i),N*L)*N*L)
ni = np.arange(-N*L/2, N*L/2)
N_i = len(xn_i)
plt.figure(figsize=(12, 8))
plt.subplot(2, 1, 1)
plt.stem(ni, xn_i)
plt.title("x_i[n]") # x[n] işaretinin grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_i[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()

#downsample işlemi
xn_d = xn_i[::M] # Downsample işlemi
N_d = len(xn_d) # Downsample sonucu oluşan dizinin uzunluğu
n_d = np.arange(0, N_d) # Downsample sonucu oluşan dizinin indisleri
# Downsample işlemi sonucu oluşan dizinin grafiğinin çizdirilmesi
plt.subplot(2, 1, 2)
plt.stem(n_d, xn_d)
plt.title("x_d[n] Downsample") # ses dosyasının grafiğinin isimlendirilmesi
plt.xlabel("örnek") # grafiğin x ekseninin isimlendirilmesi
plt.ylabel("x_d[n]") # grafiğin y ekseninin isimlendirilmesi
plt.grid()
plt.tight_layout()

plt.show() # grafiğin gösterilmesi




```
